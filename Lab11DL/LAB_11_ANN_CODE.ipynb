## Part-1 OR GATE

### get inputs and target value

```python
import numpy as np
```

```python
# inputs
x_or = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]])

# Targets
y_or = np.array([0,1,1,1])
```

### Initialize Weights

```python
weights = np.array([0,0])
lr = 0.5 #Learning rate
bias = 0

print(f"Initial Weights: {weights}")
print(f"Initial Bias: {bias}")
```

### Manual Calculation

```python
# lutput 1 if positive, 0 if ngative
def step_function(x):
    if x>=0:
        return 1
    else:
        return 0
```

```python
# Example Input: (0,1)

input_sample = x_or[1]
target_value = y_or[1]

#1. Weighted Sum (Dot Product)
z = np.dot(input_sample,weights) + bias
print(f"Math: ({input_sample[0]} * {weights[0]}) + ({input_sample[1]} * {weights[1]}) + {bias}")
print(f"Sum (z): {z:.2f}")


#2. Activation
prediction = step_function(z)
print(f"Prediction: {prediction}")

#3. Error
error = target_value - prediction
print(f"Error: {error}")
```

### The Training Loop

```python
epochs = 5

print("-------Training--------")

for epoch in range(epochs):
    print(f"\nEPOCH --> {epoch + 1}")

    for i in range(4):
        x = x_or[i]
        y = y_or[i]

        # Predict
        z = np.dot(x,weights) + bias
        pred = step_function(z)

        #Error
        error = y - pred


        # Learn (Update Weights)
        if error != 0:
            weights = weights + (lr * error * x)
            bias = bias + (lr * error * 1) # Bias Input Always One
            print(f"--> Oops!! Mistake On {x}. Expected {y}, got{pred}.")
            print(f"--> Updated Weights: {weights}, Bias: {bias:.2f}")
        else:
            print(f"Yeah!! Correct On {x}")

print("\nFinal Weights: ",weights)
print("Final Bias: ",bias)
```
