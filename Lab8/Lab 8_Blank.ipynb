# Step 7: Split Data into Training and Testing Sets
The dataset is split into training and testing sets for model evaluation.

```python
x = df.iloc[::,:-1:]
```

```python
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

# Step 8: Train Decision Tree
We train a Decision Tree Classifier on the training data. You have to also check for KNeighborsClassifier and  GaussianNB

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model =DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_predict = model.predict(x_test)
```

```python
accuracy_score(y_test,y_predict)
```

```python
model.score(x_test,y_test)
```

# Cross Validation

```python
from sklearn.model_selection import KFold
# kfold is used for the all data are as a trainig and testing and then it gives the 10 time accuracy 
#and then we will calculate all accuracy and measure average accuracy from that all accuracy and it gives th accuracy.!!
```

```python
k = KFold(n_splits=10)
```

```python
x
```

```python
for train_index,test_index in k.split(x):
    x_train,x_test = x.values[train_index],x.values[test_index]
    y_train,y_test = y[train_index],y[test_index]
    model1 = DecisionTreeClassifier()
    model1.fit(x_train,y_train)
    print(model1.score(x_test,y_test))
```

```python
from sklearn.model_selection import cross_val_score
```
